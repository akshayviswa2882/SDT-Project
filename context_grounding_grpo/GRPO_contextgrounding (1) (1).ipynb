```python
# ============================================================================
# CELL 5: Setup OpenAI API & R1 Context Grounding
# ============================================================================
%pip install openai
from openai import OpenAI

# Initialize client
client = OpenAI(api_key="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")

# R1 Rubric (GPT-4o-mini evaluates)
R1_RUBRIC = {
    "addresses_concern": 0.40,      # Does T address C's stated concern?
    "grounds_in_dialogue": 0.30,    # Is it grounded in what C said?
    "no_hallucination": 0.20,       # No invented facts?
    "no_abrupt_switching": 0.10,    # Maintains conversational flow?
}

def compute_r1_context_grounding(context, therapist_utt):
    """
    R1: Score how well the therapist utterance is grounded in context
    Returns: continuous score in [0, 1]
    """
    prompt = f"""You are an MI therapy evaluator. Rate this therapist response.

CONVERSATION HISTORY:
{context}

THERAPIST RESPONSE:
{therapist_utt}

---

EVALUATION RUBRIC (rate each 0-5):

1. ADDRESSES CLIENT'S CONCERN (weight: 0.40)
   - Does the therapist address what the client just expressed?
   - 5: Directly addresses main concern
   - 3: Partially addresses concern
   - 0: Ignores or changes topic

2. GROUNDED IN DIALOGUE (weight: 0.30)
   - Is the response based on specific details from the conversation?
   - 5: References specific client statements/concerns
   - 3: Generally related to conversation
   - 0: Generic or disconnected

3. NO HALLUCINATED FACTS (weight: 0.20)
   - Does therapist invent facts not mentioned by client?
   - 5: No invented details, only discusses what was said
   - 3: One or two assumed details
   - 0: Multiple false assumptions

4. NO ABRUPT TOPIC SWITCHING (weight: 0.10)
   - Does it maintain conversational flow?
   - 5: Natural progression from last exchange
   - 3: Some topic shift but understandable
   - 0: Abrupt, confusing topic change

---

RESPOND WITH 4 NUMBERS (0-5 each):
Example: 4 3 5 4
Meaning: addresses=4, grounds=3, hallucination=5, switching=4"""

    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        max_tokens=10,
        temperature=0.1,
        messages=[{"role": "user", "content": prompt}]
    )

    text = response.choices[0].message.content.strip()

    # Parse 4 scores
    try:
        scores = [int(x) for x in text.split()[:4]]
        if len(scores) < 4:
            scores += [2.5] * (4 - len(scores))  # default to middle
    except:
        scores = [2.5, 2.5, 2.5, 2.5]  # default if parse fails

    # Compute weighted score
    criteria = ["addresses_concern", "grounds_in_dialogue", "no_hallucination", "no_abrupt_switching"]
    weighted_sum = sum(score * R1_RUBRIC[criteria[i]] for i, score in enumerate(scores[:4]))

    # Normalize to [0, 1] (scores are 0-5, weights sum to 1.0)
    final_score = weighted_sum / 5.0

    return final_score

print("✓ OpenAI API v1.0+ ready")
print("✓ R1 Context Grounding loaded")
```

```python
# ============================================================================
# CELL 6: Build GRPO Training Data (with GT talk_type)
# ============================================================================
print("Building GRPO prompts...")

grpo_prompts = []

for tid in sorted(list(grpo_transcripts)):
    tid_df = grpo_df[grpo_df['transcript_id'] == tid].sort_values('utterance_id')

    for idx, row in tid_df.iterrows():
        # Only process client turns
        if row['interlocutor'] != 'client':
            continue

        uid = row['utterance_id']

        # Skip if no prior therapist turn
        if uid == 0:
            continue

        # Build context up to this client turn (include the client turn)
        context_df = tid_df[tid_df['utterance_id'] <= uid].sort_values('utterance_id')

        context_lines = []
        for _, ctx_row in context_df.iterrows():
            speaker = "Therapist" if ctx_row['interlocutor'] == "therapist" else "Client"
            context_lines.append(f"{speaker}: {ctx_row['utterance_text']}")

        context_text = "\n".join(context_lines)

        # Get GT: the NEXT client turn's talk_type
        next_client_df = tid_df[(tid_df['interlocutor'] == 'client') & (tid_df['utterance_id'] > uid)]

        if len(next_client_df) > 0:
            gt_talk_type = next_client_df.iloc[0]['client_talk_type']
        else:
            gt_talk_type = 'neutral'  # Default if no next client turn

        grpo_prompts.append({
            "prompt": context_text,
            "transcript_id": tid,
            "utterance_id": uid,
            "gt_talk_type": gt_talk_type  # ADD THIS
        })

print(f"Created {len(grpo_prompts)} GRPO prompts")
print(f"Example prompt with gt_talk_type: {grpo_prompts[0]}")
```

```python
# ============================================================================
# CELL 7: GRPO Generation & Scoring with R1 Only
# ============================================================================
print("Generating responses and computing R1 rewards...")

GRPO_BATCH_SIZE = 4
NUM_PROMPTS_TO_TRAIN = 50
K = GRPO_BATCH_SIZE

grpo_training_data = []

for i, prompt_dict in enumerate(grpo_prompts[:NUM_PROMPTS_TO_TRAIN]):
    prompt = prompt_dict['prompt']
    gt_talk_type = prompt_dict['gt_talk_type']  # For reference only

    # Add instruction
    full_prompt = (
        "You are a motivational interviewing therapist. "
        "Generate the next therapist response given the conversation history. "
        "Respond as the therapist only.\n\n"
        f"Conversation History:\n{prompt}\n\nGenerate the next therapist response."
    )

    # Tokenize
    inputs = tokenizer(
        full_prompt,
        return_tensors='pt',
        truncation=True,
        max_length=1024
    ).to(policy_model.device)

    # Generate K responses
    with torch.no_grad():
        outputs = policy_model.generate(
            **inputs,
            max_new_tokens=60,
            num_return_sequences=K,
            temperature=0.7,
            top_p=0.9,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id
        )

    # Decode responses
    responses = []
    for output in outputs:
        generated_tokens = output[inputs.input_ids.shape[1]:]
        response = tokenizer.decode(generated_tokens, skip_special_tokens=True).strip()

        # Clean: remove client dialogue if present
        if "Client:" in response:
            response = response.split("Client:")[0].strip()

        responses.append(response)

    # Score K responses using R1 (Context Grounding)
    rewards = []

    for response in responses:
        # R1: Context grounding score [0, 1]
        r1_score = compute_r1_context_grounding(prompt, response)
        rewards.append(r1_score)

    # Compute advantages
    mean_reward = np.mean(rewards)
    advantages = np.array([r - mean_reward for r in rewards])

    # Store training sample
    grpo_training_data.append({
        "prompt": prompt,
        "responses": responses,
        "rewards": np.array(rewards),
        "advantages": advantages,
        "gt_talk_type": gt_talk_type,  # For reference
        "mean_reward": mean_reward
    })

    if (i + 1) % 10 == 0:
        print(f"Processed {i + 1}/{NUM_PROMPTS_TO_TRAIN} prompts")
        print(f"  Mean reward: {mean_reward:.3f}")
        print(f"  Min/Max: {min(rewards):.3f} / {max(rewards):.3f}")

print(f"\nGRPO training data prepared: {len(grpo_training_data)} batches")
```

```python
# CELL 11: GRPO Training with TRL (LLM Oracle)
from trl import GRPOTrainer, GRPOConfig
from datasets import Dataset

print("Preparing GRPO dataset...")

# Convert grpo_training_data to TRL format
grpo_train_list = []
for batch in grpo_training_data:
    for response, reward in zip(batch['responses'], batch['rewards']):
        grpo_train_list.append({
            'prompt': batch['prompt'],
            'completion': response,
        })

grpo_dataset = Dataset.from_list(grpo_train_list)
print(f"GRPO dataset size: {len(grpo_dataset)}")

# Define reward function using LLM oracle
def compute_rewards(completions, **kwargs):
    """Score completions using R1 Context Grounding"""
    prompts = kwargs.get('prompts', [])
    rewards = []

    for prompt, completion in zip(prompts, completions):
        r1_score = compute_r1_context_grounding(prompt, completion)
        rewards.append(r1_score)

    return rewards


grpo_config = GRPOConfig(
    output_dir="/root/grpo_output",
    per_device_train_batch_size=8,
    num_train_epochs=2,
    learning_rate=1e-5,
    bf16=True,
    logging_steps=5,
    save_steps=50,
    num_generations=4,
    max_completion_length=60,
    temperature=0.7,
    top_p=0.9,
)


trainer = GRPOTrainer(
    model=policy_model,
    args=grpo_config,
    train_dataset=grpo_dataset,
    reward_funcs=[compute_rewards],
)

print("\n" + "="*80)
print("STARTING GRPO TRAINING")
print("="*80)

trainer.train()

print("\n" + "="*80)
print("GRPO TRAINING COMPLETE")
print("="*80)
```
